```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
from google.colab import files
uploaded = files.upload()

df = pd.read_csv("traffic.csv")

print("Dataset loaded successfully")
print(df.head())
print("Original dataset shape:", df.shape)

# Convert timestamp to datetime
df["DateTime"] = pd.to_datetime(df["DateTime"])

# Sort chronologically
df = df.sort_values(["Junction", "DateTime"])
```

```python

required_records = 2000
selected_data = None
selected_junction = None

for junction in sorted(df["Junction"].unique()):
    temp = df[df["Junction"] == junction].copy()
    temp["DateTime"] = pd.to_datetime(temp["DateTime"])
    temp = temp.sort_values("DateTime")
    temp = temp.drop_duplicates("DateTime")
    temp = temp.reset_index(drop=True)

    # Find consecutive hourly records
    hourly_gap = temp["DateTime"].diff().eq(
        pd.Timedelta(hours=1)
    )

    run_id = (~hourly_gap).cumsum()

    for _, run in temp.groupby(run_id):
        if len(run) >= required_records:
            selected_data = run.iloc[:required_records].copy()
            selected_junction = junction
            break

    if selected_data is not None:
        break

if selected_data is None:
    raise ValueError(
        "No junction has 2000 consecutive hourly records."
    )

selected_data = selected_data.reset_index(drop=True)

print("Selected Junction:", selected_junction)
print("Number of records:", len(selected_data))
print("Start time:", selected_data["DateTime"].iloc[0])
print("End time:", selected_data["DateTime"].iloc[-1])

assert len(selected_data) == 2000
assert selected_data["DateTime"].diff().dropna().eq(
    pd.Timedelta(hours=1)
).all()

print("Verified: exactly 2000 consecutive hourly records.")
```

```python

# Copy the selected 2000 records
data = selected_data.copy()

# Convert vehicle count to numeric
data["Vehicles"] = pd.to_numeric(
    data["Vehicles"], errors="coerce"
)

# Remove missing vehicle counts
data = data.dropna(subset=["Vehicles"]).reset_index(drop=True)

# Extract time-based features
data["Hour"] = data["DateTime"].dt.hour
data["DayOfWeek"] = data["DateTime"].dt.dayofweek

# Encode daily and weekly patterns
data["Hour_sin"] = np.sin(
    2 * np.pi * data["Hour"] / 24
)
data["Hour_cos"] = np.cos(
    2 * np.pi * data["Hour"] / 24
)

data["Day_sin"] = np.sin(
    2 * np.pi * data["DayOfWeek"] / 7
)
data["Day_cos"] = np.cos(
    2 * np.pi * data["DayOfWeek"] / 7
)

print("Preprocessing completed successfully!")
print("Number of records:", len(data))

print(data[[
    "DateTime",
    "Vehicles",
    "Hour",
    "DayOfWeek",
    "Hour_sin",
    "Hour_cos"
]].head())
```

```python

# Split: first 80% for training
split_index = int(len(data) * 0.8)

# Normalize vehicle counts using training data only
scaler = MinMaxScaler()

scaler.fit(data[["Vehicles"]].iloc[:split_index])

vehicles_scaled = scaler.transform(
    data[["Vehicles"]]
).flatten()

# Prepare time features
time_features = data[
    ["Hour_sin", "Hour_cos", "Day_sin", "Day_cos"]
].to_numpy()

# Combine traffic volume and time features
features = np.column_stack([
    vehicles_scaled,
    time_features
])

# Create 24-hour input windows
window_size = 24

X, y, target_indices = [], [], []

for i in range(window_size, len(data)):
    X.append(features[i-window_size:i])
    y.append(vehicles_scaled[i])
    target_indices.append(i)

X = np.array(X)
y = np.array(y)
target_indices = np.array(target_indices)

# Chronological train-test split
train_mask = target_indices < split_index
test_mask = target_indices >= split_index

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("Sequential windows created successfully!")
print("Total input windows:", len(X))
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Build LSTM model
model = Sequential([
    LSTM(
        64,
        return_sequences=True,
        input_shape=(24, 5)
    ),
    Dropout(0.2),

    LSTM(32),
    Dropout(0.2),

    Dense(16, activation="relu"),
    Dense(1)
])

# Compile the model
model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

# Display model architecture
model.summary()

# Stop training if validation loss does not improve
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

# Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)

print("LSTM model training completed!")
```

```python

# Predict traffic volume for test data
predicted_scaled = model.predict(X_test)

# Convert normalized values back to vehicle counts
actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
).flatten()

predicted = scaler.inverse_transform(
    predicted_scaled
).flatten()

# Calculate evaluation metrics
mae = mean_absolute_error(actual, predicted)

rmse = np.sqrt(
    mean_squared_error(actual, predicted)
)

print("TRAFFIC VOLUME PREDICTION RESULTS")
print("---------------------------------")
print(f"MAE  : {mae:.2f} vehicles")
print(f"RMSE : {rmse:.2f} vehicles")

# Compare actual and predicted values
results = pd.DataFrame({
    "DateTime": data["DateTime"].iloc[target_indices[test_mask]].values,
    "Actual Traffic": actual,
    "Predicted Traffic": predicted
})

print("\nFirst 10 predictions:")
print(results.head(10).round(2))
```

```python

plt.figure(figsize=(14, 6))

plt.plot(
    results["DateTime"],
    results["Actual Traffic"],
    label="Actual Traffic",
    linewidth=2
)

plt.plot(
    results["DateTime"],
    results["Predicted Traffic"],
    label="Predicted Traffic",
    linewidth=2
)

plt.title("Actual vs Predicted Hourly Traffic Volume")
plt.xlabel("Date and Time")
plt.ylabel("Number of Vehicles")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Extract hour from the prediction timestamps
results["Hour"] = pd.to_datetime(
    results["DateTime"]
).dt.hour

# Calculate average predicted traffic for each hour
hourly_forecast = results.groupby("Hour")[
    "Predicted Traffic"
].mean()

# Plot average traffic by hour
plt.figure(figsize=(10, 5))
hourly_forecast.plot(kind="bar")

plt.title("Average Predicted Traffic by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Predicted Number of Vehicles")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.tight_layout()
plt.show()

# Display the five busiest hours
busiest_hours = hourly_forecast.sort_values(
    ascending=False
).head(5)

print("Top 5 busiest hours:")
print(busiest_hours.round(2))

print("\nTraffic Planning Applications:")
print("1. Adjust traffic signal timings during peak hours.")
print("2. Allocate traffic police to busy junctions.")
print("3. Plan alternative routes during congestion.")
print("4. Monitor actual traffic against predictions.")
```
